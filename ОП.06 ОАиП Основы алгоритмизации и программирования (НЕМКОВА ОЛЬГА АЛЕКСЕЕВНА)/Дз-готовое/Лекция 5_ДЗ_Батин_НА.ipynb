````python
# Задание 1. Расчет прямоугольника
print("=== Расчет прямоугольника ===")

# Ввод данных (input всегда возвращает строку, оборачиваем в float для получения числа)
a = float(input("Введите длину прямоугольника (в см): "))
b = float(input("Введите ширину прямоугольника (в см): "))

# Вычисления
perimeter = (a + b) * 2
area = a * b

# Вывод результатов
print("\n--- Результат ---")
print("Длина:", a, "см")
print("Ширина:", b, "см")
print("Периметр:", perimeter, "см")
print("Площадь:", area, "кв.см")
````

````python
# Задание 2. Среднее арифметическое
print("=== Поиск среднего арифметического ===")

# Ввод данных
num1 = float(input("Введите первое число: "))
num2 = float(input("Введите второе число: "))
num3 = float(input("Введите третье число: "))

# Вычисление
average = (num1 + num2 + num3) / 3

# Вывод
print("\n--- Результат ---")
print("Среднее арифметическое чисел", num1, num2, num3, "равно:", average)
````

````python
# Задание 3. Магазин
print("=== Расчет стоимости покупки ===")

# Цены товаров
price_pen = 9.5
price_pencil = 7.0

# Ввод количества
pens = int(input("Сколько ручек вы купили? "))
pencils = int(input("Сколько карандашей вы купили? "))

# Расчет стоимости
total = (price_pen * pens) + (price_pencil * pencils)

# Вывод
print("\n--- Чек ---")
print("Ручки:", pens, "шт. по цене", price_pen, "руб.")
print("Карандаши:", pencils, "шт. по цене", price_pencil, "руб.")
print("ИТОГО к оплате:", total, "руб.")
````

````python
# Мой конвертер времени
print("Конвертер секунд")

# 1. Получаем данные
total_seconds = int(input("Введите количество секунд: "))

# 2. Вычисления
# Просто делим, чтобы получить дробное число часов
hours = total_seconds / 3600

# Находим остаток секунд после вычитания целых часов (задание со звездочкой)
# Оператор % (остаток от деления) покажет, сколько секунд "не влезло" в целые часы
remaining_seconds = total_seconds % 3600

# Дополнительно переведем остаток в минуты и секунды
minutes = remaining_seconds / 60
final_seconds = remaining_seconds % 60

# 3. Вывод
print("В", total_seconds, "секундах содержится:", hours, "часов.")

# Вывод остатка секунд (задание со звездочкой)
print("Остаток секунд:", int(remaining_seconds), "сек.")

# Красивый итоговый вывод (часы, минуты, секунды)
print(f"Или примерно: {int(hours)} ч. {int(minutes)} мин. {int(final_seconds)} сек.")
````

**1. Для чего мы пишем float(input(...))? Что будет, если написать просто input(...) и попытаться умножить?**

Мы пишем float(input(...)), потому что функция input всегда возвращает строку. Чтобы Python понял, что введенные данные — это число, и мог с ним математически работать, строку нужно преобразовать в число с плавающей точкой. Если написать просто input и попытаться умножить результат, Python либо продублирует строку несколько раз, либо выдаст ошибку, если умножать на дробное число.

**2. Чем отличается оператор / от оператора %? Приведите пример.**

Оператор слэш используется для обычного деления и возвращает частное, причем в Python результат всегда получается дробным. Оператор процент используется для деления по модулю и возвращает остаток от деления одного числа на другое, показывая, какая часть осталась после выделения целого.

```
a = 10
b = 3

print(a / b)  # Вывод: 3.3333333333333335 (частное)
print(a % b)  # Вывод: 1 (остаток, так как 10 = 3*3 + 1)

# Еще пример:
print(10 / 2) # Вывод: 5.0
print(10 % 2) # Вывод: 0 (остатка нет, разделилось нацело)
```


**3. В задании про магазин мы использовали int() для количества ручек. Почему мы не использовали float()?**

Для количества ручек мы используем int, потому что это целочисленная величина. В реальности нельзя купить дробное количество ручек. Float позволил бы ввести дробное число, что нарушило бы логику программы. Int гарантирует, что пользователь введет именно целое число.
