```python
# Запрашиваем у пользователя целое положительное число N
N = int(input("Введите число N: "))

# Инициализируем переменную-счетчик
i = 1

# Запускаем цикл while, который будет выполняться, пока счетчик меньше или равен N
while i <= N:
    # Выводим текущее значение счетчика на экран (с новой строки, так как print по умолчанию добавляет \n)
    print(i)

    # Увеличиваем счетчик на 1, чтобы избежать бесконечного цикла
    i += 1
```

```python
import random

# 1. Программа загадывает секретное число от 1 до 100
secret = random.randint(1, 100)
print("Компьютер загадал число от 1 до 100. Попробуйте угадать!")

# 2. Создаем переменную-счетчик попыток
attempts = 0

# 3. Организуем бесконечный цикл
while True:
    # 4. Увеличиваем счетчик попыток на 1
    attempts += 1

    # Запрашиваем у пользователя число
    guess = int(input("Ваш вариант: "))

    # Сравниваем введенное число с секретным
    if guess < secret:
        print("Слишком мало!")
    elif guess > secret:
        print("Слишком много!")
    else:
        # Числа равны
        print("Поздравляю! Вы угадали!")
        print("Количество попыток:", attempts)
        # Завершаем цикл
        break
```
